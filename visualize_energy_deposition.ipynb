```python
import pandas
import numpy as np
import matplotlib.pyplot as plt
from mpl_interactions import zoom_factory
from mpl_toolkits.mplot3d import Axes3D
from scipy.stats import gaussian_kde
%matplotlib widget
```

```python
def draw_boundaries(ax):
    # First, draw vacuum
    # vacuum dimensions: r = 274.5, h = 1758, z = 1665 mm
    vacuum_theta = np.linspace(0, 2*np.pi, 20)
    vacuum_z = np.array([1665-1758/2, 1665+1758/2])
    vacuum_theta, vacuum_z = np.meshgrid(vacuum_theta, vacuum_z)
    vacuum_x = 274.5 * np.cos(vacuum_theta)
    vacuum_y = 274.5 * np.sin(vacuum_theta)

    ax.plot_wireframe(vacuum_x, vacuum_y, vacuum_z, cmap='viridis', linestyle="--", color='black')

    # Second, draw plates
    # 50K dimensions: r = 267.5, h = 10, z = 2418 mm
    _50K_theta = np.linspace(0, 2*np.pi, 20)
    _50K_z = np.array([2418-10/2, 2418+10/2])
    _50K_theta, _50K_z = np.meshgrid(_50K_theta, _50K_z)
    _50K_x = 267.5 * np.cos(_50K_theta)
    _50K_y = 267.5 * np.sin(_50K_theta)

    ax.plot_surface(_50K_x, _50K_y, _50K_z, cmap='viridis', color='gold', alpha=0.2, edgecolor='black')
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_50K_z[0], -_50K_x, -_50K_y, np.ones(20)*_50K_z[0], color='gold', alpha=0.2)
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_50K_z[1], -_50K_x, -_50K_y, np.ones(20)*_50K_z[1], color='gold', alpha=0.2)
    
    # 4K dimensions: r = 246, h = 10, z = 2203
    _4K_theta = np.linspace(0, 2*np.pi, 20)
    _4K_z = np.array([2203-10/2, 2203+10/2])
    _4K_theta, _4K_z = np.meshgrid(_4K_theta, _4K_z)
    _4K_x = 246 * np.cos(_4K_theta)
    _4K_y = 246 * np.sin(_4K_theta)

    ax.plot_surface(_4K_x, _4K_y, _4K_z, cmap='viridis', color='gold', alpha=0.2, edgecolor='black')
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_4K_z[0], -_4K_x, -_4K_y, np.ones(20)*_4K_z[0], color='gold', alpha=0.2)
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_4K_z[1], -_4K_x, -_4K_y, np.ones(20)*_4K_z[1], color='gold', alpha=0.2)
    
    # 1k dimensions: r = 216, h = 8, z = 2027
    _1K_theta = np.linspace(0, 2*np.pi, 20)
    _1K_z = np.array([2027-8/2, 2027+8/2])
    _1K_theta, _1K_z = np.meshgrid(_1K_theta, _1K_z)
    _1K_x = 216 * np.cos(_1K_theta)
    _1K_y = 216 * np.sin(_1K_theta)

    ax.plot_surface(_1K_x, _1K_y, _1K_z, cmap='viridis', color='gold', alpha=0.2, edgecolor='black')
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_1K_z[0], -_1K_x, -_1K_y, np.ones(20)*_1K_z[0], color='gold', alpha=0.2)
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_1K_z[1], -_1K_x, -_1K_y, np.ones(20)*_1K_z[1], color='gold', alpha=0.2)
    
    # 100mK dimensions: r = 192.5, h = 8, z = 1850
    _100mK_theta = np.linspace(0, 2*np.pi, 20)
    _100mK_z = np.array([1850-8/2, 1850+8/2])
    _100mK_theta, _100mK_z = np.meshgrid(_100mK_theta, _100mK_z)
    _100mK_x = 192.5 * np.cos(_100mK_theta)
    _100mK_y = 192.5 * np.sin(_100mK_theta)

    ax.plot_surface(_100mK_x, _100mK_y, _100mK_z, cmap='viridis', color='gold', alpha=0.2, edgecolor='black')
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_100mK_z[0], -_100mK_x, -_100mK_y, np.ones(20)*_100mK_z[0], color='gold', alpha=0.2)
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_100mK_z[1], -_100mK_x, -_100mK_y, np.ones(20)*_100mK_z[1], color='gold', alpha=0.2)
    
    # MCP1 dimensions: r = 180, h = 8, z = 1600
    _MCP1_theta = np.linspace(0, 2*np.pi, 20)
    _MCP1_z = np.array([1600-8/2, 1600+8/2])
    _MCP1_theta, _MCP1_z = np.meshgrid(_MCP1_theta, _MCP1_z)
    _MCP1_x = 180 * np.cos(_MCP1_theta)
    _MCP1_y = 180 * np.sin(_MCP1_theta)

    ax.plot_surface(_MCP1_x, _MCP1_y, _MCP1_z, cmap='viridis', color='gold', alpha=0.2, edgecolor='black')
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_MCP1_z[0], -_MCP1_x, -_MCP1_y, np.ones(20)*_MCP1_z[0], color='gold', alpha=0.2)
    ax.fill_between(np.zeros(20), np.zeros(20), np.ones(20)*_MCP1_z[1], -_MCP1_x, -_MCP1_y, np.ones(20)*_MCP1_z[1], color='gold', alpha=0.2)
    
    return ax
```

```python
dataframe = pandas.read_csv('QUIETFridgeCoarseOutput_nt_QUIETFridgeCoarse_t0.csv')

x = dataframe.loc[:, 'startPosition_x'].to_numpy()
y = dataframe.loc[:, 'startPosition_y'].to_numpy()
z = dataframe.loc[:, 'startPosition_z'].to_numpy()
xyz = np.vstack([x, y, z])
density = gaussian_kde(xyz)(xyz)

idx = density.argsort()
x, y, z, density = x[idx], y[idx], z[idx], density[idx]
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, c=density)

draw_boundaries(ax)

# Enable scroll to zoom with zoom_factory
disconnect_zoom = zoom_factory(ax, base_scale=1.02)

plt.show()
```
