```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
from attenuation_coefficients import gamma_energies_keV, attenuation_factors
```

```python
# Read NumPy data
infile_dir = 'graph_data'
data = np.load(infile_dir + '/verification_data.npz')
gamma_energies = np.arange(0, 30) * 100

print("Successfully loaded data!")
```

```python
# Attenuation Factor vs. Gamma Energy
fig, ax = plt.subplots()

ax.scatter(gamma_energies, data['ngamma'] / 1000000 * 100, label='Actual', color='blue')
ax.plot(gamma_energies_keV, attenuation_factors * 100, label='Predicted', color='green')
ax.legend()

# Axis labels
ax.set_xlabel('Gamma Energy (keV)')
ax.set_ylabel('Ratio of Gammas Reaching Chip (%)')
ax.set_title('Attenuation Factor vs. Gamma Energy')

plt.show()
```
