```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read NumPy data
infile_dir = 'graph_data'
angles = np.array([0, 15, 30, 45, 60, 75, 90, 105, 120, 135, 150, 165, 180, 195, 210, 225, 240, 255, 270, 285, 300, 315, 330, 345])
data_OLAF = np.load(infile_dir + '/OLAF_angular_data.npz')
data_QUIET = np.load(infile_dir + '/QUIET_angular_data.npz')

print("Successfully loaded data!")
```

```python
N_OLAF = 4000000000
N_QUIET = 4000000000
```

```python
# Share of Events with Gammas vs. Angle
plt.scatter(angles, data_OLAF['ngamma'] / N_OLAF * 100, label="OLAF", color='#f94449')
plt.errorbar(angles, data_OLAF['ngamma'] / N_OLAF * 100, yerr=data_OLAF['ngamma_std'] * 100, linestyle="None", color='#f94449')

plt.scatter(angles, data_QUIET['ngamma'] / N_QUIET * 100, label="QUIET", color='#0496c7')
plt.errorbar(angles, data_QUIET['ngamma'] / N_QUIET * 100, yerr=data_QUIET['ngamma_std'] * 100, linestyle="None", color='#0496c7')

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Percentage of Events with Gammas (%)')
plt.title('Gamma Flux vs. Angle')
plt.legend()

plt.show()
```

```python
# Share of Events with Energy Deposition vs. Angle

plt.scatter(angles, data_OLAF['energydepevts'] / N_OLAF * 100, label="OLAF", color='#f94449')
plt.errorbar(angles, data_OLAF['energydepevts'] / N_OLAF * 100, yerr=data_OLAF['energydepevts_std'] * 100, linestyle="None", color='#f94449')

plt.scatter(angles, data_QUIET['energydepevts'] / N_QUIET * 100, label="QUIET", color='#0496c7')
plt.errorbar(angles, data_QUIET['energydepevts'] / N_QUIET * 100, yerr=data_QUIET['energydepevts_std'] * 100, linestyle="None", color='#0496c7')

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Percentage of Events with Energy Deposition (%)')
plt.title('Share of Events with Energy Deposition vs. Angle')
plt.legend()

plt.show()
```

```python
# Energy Deposited in Chip per Event vs. Angle

plt.scatter(angles, data_OLAF['energydep'] / data_OLAF['energydepevts'], label="OLAF", color='#f94449')
plt.errorbar(angles, data_OLAF['energydep'] / data_OLAF['energydepevts'], yerr=data_OLAF['energydep_std'], linestyle="None", color='#f94449')

plt.scatter(angles, data_QUIET['energydep'] / data_QUIET['energydepevts'], label="QUIET", color='#0496c7')
plt.errorbar(angles, data_QUIET['energydep'] / data_QUIET['energydepevts'], yerr=data_QUIET['energydep_std'], linestyle="None", color='#0496c7')

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Energy Deposited (keV)')

plt.title('Energy Deposited in Chip per Event vs. Angle')
plt.legend()

plt.show()
```

```python
cs137_activity = 7.26e-6
decays_per_second = 3.7e10
```

```python
# Share of Events with Gammas vs. Angle (OLAF)
plt.scatter(angles, cs137_activity * decays_per_second * data_OLAF['ngamma'] / N_OLAF, color='#f94449', label='OLAF')
plt.errorbar(angles, cs137_activity * decays_per_second * data_OLAF['ngamma'] / N_OLAF, yerr=cs137_activity * decays_per_second * data_OLAF['ngamma_std'], linestyle="None", color='#f94449')

# Share of Events with Gammas vs. Angle (QUIET)
plt.scatter(angles, cs137_activity * decays_per_second * data_QUIET['ngamma'] / N_QUIET, color='#0496c7', label='QUIET')
plt.errorbar(angles, cs137_activity * decays_per_second * data_QUIET['ngamma'] / N_QUIET, yerr=cs137_activity * decays_per_second * data_QUIET['ngamma_std'], linestyle="None", color='#0496c7')

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('# Gammas per Second (1/s)')
plt.title('Gamma Flux vs. Radial Distance')
plt.legend()

plt.show()
```

```python
# Share of Events with Gammas vs. Angle (OLAF)
plt.scatter(angles, cs137_activity * decays_per_second * data_OLAF['energydep'] / N_OLAF, color='#f94449', label='OLAF')
edep_std = np.sqrt((data_OLAF['energydepevts_std']**2) * (data_OLAF['energydep_std']**2) + (data_OLAF['energydep_std']**2) * ((data_OLAF['energydepevts'] / N_OLAF)**2) + (data_OLAF['energydepevts_std']**2) * (data_OLAF['energydep'] / data_OLAF['energydepevts'])**2)
plt.errorbar(angles, cs137_activity * decays_per_second * data_OLAF['energydep'] / N_OLAF, yerr=cs137_activity * decays_per_second * edep_std, linestyle="None", color='#f94449')

# Share of Events with Gammas vs. Angle (QUIET)
plt.scatter(angles, cs137_activity * decays_per_second * data_QUIET['energydep'] / N_QUIET, color='#0496c7', label='QUIET')
edep_std = np.sqrt((data_QUIET['energydepevts_std']**2) * (data_QUIET['energydep_std']**2) + (data_QUIET['energydep_std']**2) * ((data_QUIET['energydepevts'] / N_QUIET)**2) + (data_QUIET['energydepevts_std']**2) * (data_QUIET['energydep'] / data_QUIET['energydepevts'])**2)
plt.errorbar(angles, cs137_activity * decays_per_second * data_QUIET['energydep'] / N_QUIET, yerr=cs137_activity * decays_per_second * edep_std, linestyle="None", color='#0496c7')

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Energy Deposition per Second (keV/s)')
plt.title('Energy Deposition vs. Radial Distance')
plt.legend()

plt.show()
```
