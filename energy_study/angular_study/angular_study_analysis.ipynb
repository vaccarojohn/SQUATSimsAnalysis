```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read NumPy data
infile_dir = 'graph_data'
angles = np.array([0, 15, 30, 45, 60, 75, 90, 105, 120, 135, 150, 165, 180, 195, 210, 225, 240, 255, 270, 285, 300, 315, 330, 345])
data = np.load(infile_dir + '/cs137_data.npz')

print("Successfully loaded data!")
```

```python
N = 8000000000
```

```python
# Share of Events with Gammas vs. Angle
plt.scatter(angles, data['ngamma'] / N * 100)
plt.errorbar(angles, data['ngamma'] / N * 100, yerr=data['ngamma_std'] * 100, linestyle="None")

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Percentage of Events with Gammas (%)')
plt.title('Gamma Flux vs. Angle')

plt.show()
```

```python
# Share of Events with Energy Deposition vs. Angle

plt.scatter(angles, data['energydepevts'] / N * 100)
plt.errorbar(angles, data['energydepevts'] / N * 100, yerr=data['energydepevts_std'] * 100, linestyle="None")

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Percentage of Events with Energy Deposition (%)')
plt.title('Share of Events with Energy Deposition vs. Angle')

plt.show()
```

```python
# Energy Deposited in Chip per Event vs. Angle

plt.scatter(angles, data['energydep'] / data['energydepevts'])
plt.errorbar(angles, data['energydep'] / data['energydepevts'], yerr=data['energydep_std'], linestyle="None")

# Axis labels
plt.xlim(0, 360)
plt.xticks(np.arange(0, 405, step=45)) 
plt.xlabel('Angle (°)')
plt.ylabel('Energy Deposited (keV)')

plt.title('Energy Deposited in Chip per Event vs. Angle')

plt.show()
```
