```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read NumPy data
infile_dir = 'graph_data'
data = np.load(infile_dir + '/energy_study_data.npz')
gamma_energies = np.arange(0, 30) * 100

print("Successfully loaded data!")
```

```python
# Percentage of Events with Energy Deposition vs. Gamma Energy

plt.scatter(gamma_energies, data['energydepevts'] / 1000000 * 100)

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Percentage of Events with Energy Deposition (%)')
plt.title('Percentage of Events with Energy Deposition vs. Gamma Energy')

plt.show()
```

```python
# Percentage of Events with Gamma Energy Deposition vs. Gamma Energy

plt.scatter(gamma_energies[1:], data['gammadepevts'][1:] / data['energydepevts'][1:]* 100)

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Percentage of Events with Gamma Energy Deposition (%)')
plt.title('Percentage of Energy Deposition Events with Gamma Energy Deposition vs. Gamma Energy')

plt.show()
```

```python
# Energy Deposited in Chip vs. Gamma Energy

plt.scatter(gamma_energies, data['energydep'] / 1000000)

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Energy Deposited (keV)')
plt.title('Average Energy Deposited in SQUAT vs. Gamma Energy')

plt.show()
```

```python
# Energy Deposited in Chip by Gammas vs. Gamma Energy

plt.scatter(gamma_energies, data['gammadep'] / 1000000)

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Energy Deposited (keV)')
plt.title('Average Energy Deposited in SQUAT by Gammas vs. Gamma Energy')

plt.show()
```

```python
# Energy Deposited in Chip vs. Gamma Energy

plt.scatter(gamma_energies[1:], data['energydep'][1:] / data['energydepevts'][1:])

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Energy Deposited (keV)')
plt.title('Average Energy Deposited in SQUAT (Excluding Empty Events) vs. Gamma Energy')

plt.show()
```

```python
# Energy Deposited in Chip vs. Gamma Energy
plt.scatter(gamma_energies[2:], data['gammadep'][2:] / data['gammadepevts'][2:])

# Axis labels
plt.xlabel('Gamma Energy (keV)')
plt.ylabel('Energy Deposited (keV)')
plt.title('Average Energy Deposited in SQUAT (Excluding Empty Events) vs. Gamma Energy')

plt.show()
```
